### Create a function that works the same as the strip method, so you can see how the strip method works

```python
s1 = '   fgedeef   '

def mystrip(s1, s2 = ' \n\t'):
    lindex = 0
    while lindex < len(s1) and s1[lindex] in s2:
        lindex += 1    
    rindex = -1
    while rindex >= -len(s1) and s1[rindex] in s2:
        rindex -= 1    
    return s1[lindex:rindex+1]        

print(mystrip(s1, ' fg') == s1.strip(' fg'))
            
        
```

### You can use class attributes (in the followinexample total_size) to keep track of information you want to share between objects created on the basis of the same class.

```python
class Rectangle:
    total_size = 0
    def __init__(self, length, width=1):
        self.length = length
        self.width = width
        Rectangle.total_size += self.length * self.width
        print(f'rectangle {self.length} by {self.width} created\n{Rectangle.total_size = }')
    def size(self, data_type=int):
        return f"The size = {data_type(self.length * self.width)}."
    def __del__(self):
        Rectangle.total_size -= self.length * self.width
        print(f'rectangle {self.length} by {self.width} deleted\n{Rectangle.total_size = }')
    
print(Rectangle.total_size)
rectangle_1 = Rectangle(2,3)
rectangle_2 = Rectangle(3)
rectangle_3 = Rectangle(2,2)

del rectangle_1
del rectangle_3
del rectangle_2
```

### #<br>Referencing and defining attributes from the outside

```python
#
class Rectangle:
    def __init__(self, length, width=1):
        self.length = length
        self.width = width
    def info_about_attribute_a(self):
        return f'{Rectangle.a=}, {self.a=}'

rectangle_1 = Rectangle(2,3)
Rectangle.a = 5
rectangle_1.a = 10
print(Rectangle.a, rectangle_1.a)
print(rectangle_1.info_about_attribute_a())
```

### #<br>Create a class for which all the methods work the same as in the str class except that the index method now works the same as the find method

```python
#
class New_str(str):
    def index(self, needle, start = 0):
        return self.find(needle, start)

a = New_str("abcd")
print(a.upper()) 
print(a.find("d")) 
print(a.find("e")) 
print(a.index("d")) 
print(a.index("e")) 
```

### Create a function that works like the built-in index method, but if the substring is not found, it doesn't throw an error, but returns -1

```python
def my_index(s,needle):
    try:
        result = s.index(needle)
    except:
        result = -1
    return result
s1 ='abcdefghabcdefgh'
print(my_index(s1, 'e'))
print(my_index(s1, 'i'))
```

```python
def my_find(s,needle):
    if (result := s.find(needle)) == -1:
        raise ValueError("substring not found")
    return result
s1 ='abcdefghabcdefgh'
print(my_find(s1, 'e'))
print(my_find(s1, 'i'))
```
