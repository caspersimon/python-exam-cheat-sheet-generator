### Selecting data from a dataframe using some shortcuts. The code in the comments would gives the same selection

```python
import pandas as pd
d1 = {'Name':['A', 'B', 'C'],'Length':[153, 160, 150],'Weight':[55, 70, 60]}
df1 = pd.DataFrame (d1, index=['First', 'Second', 'Third'])
print(df1)

print(df1.Name)                      # df1.loc[:,'Name']
print(df1['Name'])                   # df1.loc[:,'Name']
print(df1[['Name']])                 # df1.loc[:,['Name']]
print(df1[['Name', 'Weight']])       # df1.loc[:,['Name', 'Weight']]
print(df1.Name['First'])             # df1.loc['First','Name']
print(df1.Name[['First', 'Second']]) # df1.loc[['First', 'Second']:,'Name']
```

### Vectorized string methods

```python
import pandas as pd
s1 = pd.Series({1:'Aap', 2:'Nota', 4:'MIES'})
print(s1)

print(s1.map(lambda x: x[::-1]))
print(s1.str[::-1])

print(s1.map(lambda x: x.lower()))
print(s1.str.lower())

print(s1.map(lambda x: x.count('a')))
print(s1.str.count('a'))

print(s1.map(lambda x: x.isupper()))
print(s1.str.isupper())
```

```python
import pandas as pd
s1 = pd.Series(list('Python'))
print(s1)

print(s1.map(lambda x: x.lower() in 'aeiou'))
print(s1.str.lower().isin(list('aeiou')))
print(s1[s1.str.lower().isin(list('aeiou'))])

print(s1.map(lambda x: x.lower() not in 'aeiou'))
print(~s1.str.lower().isin(list('aeiou')))
print(s1[~s1.str.lower().isin(list('aeiou'))])
```

### Concatenating DataFrames

### Merging and grouping DataFrames

```python
import pandas as pd
dftax = pd.DataFrame({'Name':['A', 'B', 'C'], 'Income':[100, 200, 150], 'Tax':[10, 60, 30], 'City':['Amsterdam', 'Rotterdam', 'London']} )
print(dftax)

dfprov = pd.DataFrame({'City':['Amsterdam', 'Rotterdam', 'Utrecht'], 'Province': ['Noord Holland', 'Zuid Holland', 'Utrecht']})
print(dfprov)

print(dftax.merge(dfprov, left_on='City', right_on = 'City', how='left'))
print(dftax.merge(dfprov, left_on='City', right_on = 'City', how='right'))

dfcombined = dftax.merge(dfprov, how='left')
dfcombined.loc[:,'Province'] = dfcombined.loc[:,'Province'].fillna('Rest')
# We replace the missing values in the Province column with a string
# As the groupby method ignores NaN values
result = dfcombined.groupby('Province')['Tax'].sum()
result = result.sort_values(ascending = False)
# We want the highest paying Province first
print(result)

print(result[[0]])
print(result.sum())
print(result['Noord Holland'])
```
