### List comprehensions are a nice way to make your code more concise, but you can always do the same with a loop 

```python
l1 = [1, 3, 2, 4] 

l2 = []
for x in l1:
    if x%3 ==0:
        continue
    addition = x * 2 if x%2 == 1 else x // 2
    l2.append(addition)
print(l2)

l2 = [x * 2 if x%2 == 1 else x // 2 for x in l1 if x%3 != 0]
print(l2)
```

### Dictionary comprehensions 

```python
l1 = [1, 3, 2, 5]
l2 = [1, 2, 6, 3]

d1 = {}
for k,v in zip(l1,l2):
    if k%2 == 0 or v%2 == 0:
        continue
    d1[k] = v * 3
print(d1)

d1 = {k: v*3 for k,v in zip(l1,l2) if not (k%2 == 0 or v%2 == 0)}
print(d1)
```

### Set comprehensions

```python
l1 = [1, 3, 2, 5] 

s1 = set()
for x in l1:
    if x % 2 == 0:
        continue
    s1.add(x * 2)
print(s1)

s1 = {x * 2 for x in l1 if not (x%2 == 0)}
print(s1)
```

### 'String comprehensions' 

```python
s1 = 'a1BcDE2FgH3i'

import string
s2 = ''
for x in s1:
    if x in string.digits:
        continue
    s2 += x * 2 if x.lower() in 'aeiou' else x * 3
print(s2)

import string
l1 = [x * 2 if x.lower() in 'aeiou' else x * 3 for x in s1 if x in string.ascii_letters]
s2 = ''.join(l1)
print(s2)
```

### Nested comprehensions

```python
func = lambda x,y: x+y

result = {}
for x in range(1, 4):
    result1 = {}
    for y in range(x, 4):
        result1[y] = func(x,y)
    result[x] = result1    
print(result)
print(result[2][3])

del x, y
result = {x:{y: func(x,y) for y in range(x, 4)} for x in range(1, 4)}
print(result)
print(result[2][3])
```

```python
func = lambda x,y: x+y

result = {}
for x in range(1, 4):
    for y in range(x, 4):
        result[x,y] = func(x,y)
print(result)        
print(result[2,3])

del x,y
result = {(x, y): func(x,y) for x in range(1, 4) for y in range(x, 4) }
print(result)
print(result[2,3])
```

### List comprehensions (old school)

```python
l1 = [1, 3, 2, 4] 

l2 = []
for x in l1:
    if x%3 ==0:
        continue
    addition = x * 2 if x%2 == 1 else x // 2
    l2.append(addition)
print(l2)

l2 = [x * 2 if x%2 == 1 else x // 2 for x in l1 if x%3 != 0]
print(l2)

l2 = list(map(lambda x: x * 2 if x%2 == 1 else x//2, filter(lambda x: x % 3 != 0, l1)))
print(l2) 

filtered_values = filter(lambda x: x % 3 != 0, l1)
processed_values = map(lambda x: x * 2 if x%2 == 1 else x//2, filtered_values)
list_of_processed_values = list(processed_values)
print(list_of_processed_values)
```

### Keeping state between function calls with a standard function

```python
def myfunc(start, stop,step):
    num = start
    start += step
    if num > stop:
        raise StopIteration('End of numbers')
    else:
        return num, start, stop, step
sta = 1
sto  = 5
ste  = 2
num, sta, sto, ste = myfunc(sta, sto, ste)
print(num)
num, sta, sto, ste = myfunc(sta, sto, ste)
print(num)
num, sta, sto, ste = myfunc(sta, sto, ste)
print(num)
num, sta, sto, ste = myfunc(sta, sto, ste)
print(num)
```

### Keeping state between function calls with a standard function and the global keyword

```python
start = 1
stop  = 5
step  = 2
def myfunc():
    global start, stop, step 
    num = start
    start += step
    if num > stop:
        raise StopIteration('End of numbers')
    else:
        return num
start = 1
stop  = 5
step  = 2    
print(myfunc())
print(myfunc())
print(myfunc())
print(myfunc())
```

### Keeping state between function calls with a standard class

```python
class Myclass:
    def __init__(self, start, stop, step):
        self.start = start
        self.stop = stop
        self.step = step
    def following(self):
        self.num = self.start
        self.start += self.step
        if self.num > self.stop:
            raise StopIteration('End of numbers')
        else:
            return self.num
myfunc = Myclass(1, 5, 2)
print(myfunc.following())
print(myfunc.following())
print(myfunc.following())
print(myfunc.following())
```

### Keeping state between function calls with an iterator

```python
class Myiterator:
    def __init__(self, start, stop, step):
        self.start = start
        self.stop = stop
        self.step = step
    def __next__(self):
        self.num = self.start
        self.start += self.step
        if self.num > self.stop:
            raise StopIteration('End of numbers')
        else:
            return self.num
myiterator = Myiterator(1, 5, 2)
print(next(myiterator))
print(next(myiterator))
print(next(myiterator))
print(next(myiterator))
```

### Keeping state between function calls with an generator function

```python
def generatorfunction(start, stop, step):
    while (num:=start) <= stop:
        yield num
        start += step
    return 'End of numbers'
mygenerator = generatorfunction(1, 5, 2)
print(next(mygenerator))
print(next(mygenerator))
print(next(mygenerator))
print(next(mygenerator))
```

```python
def generatorfunction():
    num = 1
    yield num
    num += 2
    yield num
    num += 2
    yield num

mygenerator = generatorfunction()
print(next(mygenerator))
print(next(mygenerator))
print(next(mygenerator))
print(next(mygenerator))
```

### Turning a generatorfunction into a list

```python
def generatorfunction(start, stop, step):
    while True:
        num = start
        start += step
        if num <= stop:
            yield num
        else:
            return 'End of numbers'
print(list(generatorfunction(1, 5, 2)))
```

### Arithmetic with datetime and timedelta objects

```python
from datetime import datetime
birthday = datetime(1923, 10, 26)
age = datetime.now() - birthday

print(f"""
{age.days =}
{age.seconds =}
""")
print(type(age))
```

```python
from datetime import datetime
birthday = datetime(1923, 10, 26)
today = datetime.now()

age = today.year - birthday.year - 1
age += today.month > birthday.month or today.month == birthday.month and today.day >= birthday.day

print(f"""
{age = } years
""")
```
